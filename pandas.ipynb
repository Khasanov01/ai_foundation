```python
mydataset={
    'cars': ["BMW", "Volvo", "Ford", "Tesla", "Ferrari", "Ford", "BMW"],
    'counts': [3, 7, 2, 8, 1, 9, 3]
}
print(mydataset)
df=pd.DataFrame(mydataset)
print(df)

df.head(0) #default 5 rows

df.tail(7) #default 5 rows

df.shape

df.columns

df.dtypes

df.describe() #numerical columns only

df.iloc[2]

df.iloc[1:5]

df.loc[:, "cars"]

df[df["counts"]>3]

df[df["cars"]== "BMW"]

df.mean(numeric_only=True)

df["counts"].mean()

df.sum()

df.min()

df.max()

df.median(numeric_only=True)

df.count()

df.std(numeric_only=True)


####
data_with_missing={
    "cars":["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    "counts": [3, 7, None, 8, 1, 9, None]
}

df_missing=pd.DataFrame(data_with_missing)
print(df_missing)

df_missing.isnull().sum()

df_missing.dropna(how="all")

df_missing.dropna(subset=['counts'])

df_missing.fillna({"cars":"unknown", "counts": 0})

cars_series=data_with_missing['cars']
print(cars_series)
```
